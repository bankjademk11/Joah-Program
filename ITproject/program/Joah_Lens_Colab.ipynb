```python
# 1. ติดตั้ง Git LFS และ Clone Repository จาก Hugging Face (พร้อม Index 23,930 รายการ)
!apt-get update -qq && apt-get install git-lfs -y -qq
!git lfs install
!rm -rf /content/joah-lens-search
!git clone https://huggingface.co/spaces/bankjademk11/joah-lens-search /content/joah-lens-search
%cd /content/joah-lens-search
!git lfs pull

# 2. ติดตั้ง Python Libraries
!pip install -q fastapi "uvicorn[standard]" pydantic Pillow numpy tqdm open_clip_torch faiss-cpu rembg onnxruntime torchvision

# 3. ดาวน์โหลด Cloudflare Tunnel (สร้าง Public HTTPS URL ฟรี ไม่ต้องสมัครสมาชิก)
!wget -q -O cloudflared https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64
!chmod +x cloudflared

# 4. สตาร์ท FastAPI Server และเปิด Public Link
import subprocess, time, re

server_proc = subprocess.Popen(["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(3)

tunnel_proc = subprocess.Popen(["./cloudflared", "tunnel", "--url", "http://localhost:8000"], stderr=subprocess.PIPE, text=True)

print("⏳ กำลังเปิด Public HTTPS URL...")
public_url = None
for line in tunnel_proc.stderr:
    match = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', line)
    if match:
        public_url = match.group(0)
        break

print("\n" + "="*65)
print(f"🎉 Joah Lens Search API Online บน Google Colab GPU (T4)!")
print(f"👉 Public URL ของคุณคือ:")
print(f"   {public_url}")
print("="*65)
print("นำ URL นี้ไปใส่ในเว็บ หรือบอก AI เพื่อเชื่อมต่อได้ทันทีครับ!")
```
